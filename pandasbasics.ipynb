```python
#DATA CLEANING
```

```python
#dropping the missing values by dropna
```

```python
df.isnull().sum()
```

```python
 df.dropna()#how = "any" #meaning if even one value if null whole row is dropped
```

```python
df.dropna(how = "all")#meaning if all values of the row are null then only the row is dropped
```

new topic

```python
#filling the missing values by fillna
```

```python
df.fillna(0)#worst method
```

```python
df["Price"].fillna(df["Price"].mean())
```

```python
df["Rating"].fillna(df["Rating"].median())
```

```python
#filling value using forwardfill -> upper value will be filled, backwardfill -> lower value will be filled in the null value
```

```python
df["Price"].fillna(method = "ffill")
```

```python
df["Price"].fillna(method = "bfill")
```

```python
df["Product"].replace("Tablet", "Ipad")
df["Product"].replace("Laptop", "Macbook")
```

```python
df_dup = df[df.duplicated()] #same as df_dup = df[df.duplicated(keep = "first")]
df_dup
```

```python
df_dup = df[df.duplicated(keep = "last")]
df_dup
```

```python
df = df.drop_duplicates()
df
```
