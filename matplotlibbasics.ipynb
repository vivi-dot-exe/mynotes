```python
# 1. Update data with some realistic "bumps" (fluctuations)
df["age"] =    [22,    25,    27,    29,    31,    32,    35,    38,    40,    42,    45,    48]
df["Salary"] = [52000, 61000, 48000, 55000, 75000, 60000, 68000, 82000, 71000, 50000, 79000, 84000]

# 2. Sort it by age so it stays left-to-right
sort_age = df.sort_values("age")

# 3. Plot it again
plt.plot(sort_age["age"], sort_age["Salary"], color="pink", marker="o", linewidth=1)
plt.grid(True)
plt.show()
```

```python
#BAR CHART
```

```python
plt.bar(df["age"], df["Salary"], color = "pink" )
plt.show()
```

```python
plt.bar(df["age"], df["Salary"], color = "pink" , align = "center")
plt.show()
```

```python
#BIVARIATE
```

```python
#NUMERICAL - CATEGORICAL
```

```python
hr_salary = df[df["dept"] == "HR"]["Salary"]
hr_salary
```

```python
it_salary = df[df["dept"] == "IT"]["Salary"]
it_salary
```

```python
finance_salary = df[df["dept"] == "Finance"]["Salary"]
finance_salary
```

```python
plt.boxplot([hr_salary, it_salary, finance_salary], labels = ["HR", "IT", "Finance"])
plt.show()
```

```python
#PieChart
```

```python
salary_by_dept = df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept, labels = salary_by_dept.index, autopct = "%1.1f%%", shadow = True)
plt.axis("equal")
plt.show()
```

```python
#BAR PLOT
```

```python
hr_mean = sum(hr_salary)/len(hr_salary)
it_mean = sum(it_salary)/len(it_salary)
finance_mean = sum(finance_salary)/len(finance_salary)
```

```python
plt.bar(["HR", "IT", "Finance"], [hr_mean, it_mean, finance_mean], color = ['pink', 'red', 'maroon'])
plt.grid()
plt.show()
```

```python
df.head()
```

```python
df["experience"] = [1,2,4,1.5,5,8,12,19,20,1,3,4]
df.head(12)
```

```python
#BUBBLE PLOT
```

```python
plt.scatter(df['age'], df['Salary'], s=df['experience'], color="maroon")
plt.show()
```

```python
plt.scatter(df['age'], df['Salary'], s=df['experience']*5 , color="pink", edgecolor="maroon")
plt.title("Age vs Salary vs Experience")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid()
plt.show()
```

```python
#two numerical one categorical
```

```python
plt.scatter(df['age'], df['Salary'], c = df['dept'].map({"HR": "blue", "IT": "skyblue", "Finance": "purple"}))
plt.legend()
plt.title("Age vs Salary vs Department")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
color ={"HR": "blue", "IT": "skyblue", "Finance": "purple"}
for dept, color in color.items():
    df_dept =df[df["dept"] == dept]
    plt.scatter(df_dept["age"], df_dept["Salary"], c=color, label = dept)
plt.legend()
```
